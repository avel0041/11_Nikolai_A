```python
res=[]
for n in range(1,1000):
    b=bin(n)[2:]
    if b.count('1')%2==0:
        b='1'+b[2:]+'0'
    else:
        b='11'+b[2:]+'1'
    if int(b,2)==742:
        res.append(n)
print(res)

```

```python
# print(bin(903)[2:])
# # 1110000111
# # 1011111110
# print(int('1011111110',2))

res=[]

for n in range(1, 10**6):
    i = 5
    s = list('*****' + bin(n)[2:] + '*****')
    print(s)
    while s[i]!='*':
        i+=1
    s[i] = '0'
    i-=1
    while s[i]!='*':
        if s[i]=='0':
            s[i]= '1'
        i-=1
    s[i] = '1'
    i -= 1
    s[i] = '0'
    i -= 1
    s[i] = '1'
    i -= 1
    st = ''
    for z in s:
        if z!='*':
            st = st + z
    if int(st,2)<=903:
        res.append(int(st,2))
print(max(res))
```
