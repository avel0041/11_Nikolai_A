```python
# 186
f = open('24-181.txt').readline()
c=0
for i in range(len(f)-10):
    n = f[i:i+11]
    f1=0
    for x in n:
        if x not in '0123456789':
            f1=1
    if f1 == 0:
        if n[0]=='7' and (int(n[0])+int(n[1]))==(int(n[-1])+int(n[-2])):
            c+=1
print(c)
```

```python
# 189
f = open('24-181.txt').readline()
f = f.replace('A', 'Y')
f = f.replace('E', 'Y')
f = f.replace('I', 'Y')
f = f.replace('O', 'Y')
f = f.replace('U', 'Y')


k=0
v = 0
l = 0
b = 0
for i in range(len(f)):
    if f[i]=='Y':
        v+=1
    if f[i]=='.':
        b+=1
    while v>7 or b>0:
        if f[l]=='Y':
            v-=1
        if f[l]=='.':
            b-=1
        l+=1
    if (i-l)>k:
        k=i-l
print(k+1)
```

```python
# 190
f = open('24-181.txt').readline()
f = f.replace('A', 'Y')
f = f.replace('E', 'Y')
f = f.replace('I', 'Y')
f = f.replace('O', 'Y')
f = f.replace('U', 'Y')

f = f.split('Y')
mx = 0

for el in f:
    if el.count('.')>=6:
        mx = max(len(el), mx)

print(mx)
```

```python
f = open('24.txt').readline().replace('2', '0').replace('4', '0').replace('6', '0').replace('8', '0')
m = 1

for l in range(len(f)):
    for r in range(l+m, len(f)):
        s = f[l:r+1]
        if s[0] == '0' and s.count('0') == 1:
            if s.count('W')==36:
                m = max(m, len(s))
        else:
            break
print(m)
```

```python
f=open('241.txt').readline()
mx = 52
st = ''
for s in 'EIOU':
    f=f.replace(s,'*')
for l in range(len(f)-1):
    if f[l] != 'A':
        continue
    for r in range(l+mx, len(f)):
        if f[r] != 'Z':
            continue
        s = f[l:r+1]
        su = 0
        for el in s:
            if el in '123456789':
                su += int(el)
        if (s.count('*')+s.count('A'))==50 and su%7==0 and s.count('Z')==1:
            z=len(s)
            if z>mx:
                mx=z
                st=s
        else:
            break
print(mx)
print(st)
```
