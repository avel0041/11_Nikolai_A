```python
# 186
f = open('24-181.txt').readline()
c=0
for i in range(len(f)-10):
    n = f[i:i+11]
    f1=0
    for x in n:
        if x not in '0123456789':
            f1=1
    if f1 == 0:
        if n[0]=='7' and (int(n[0])+int(n[1]))==(int(n[-1])+int(n[-2])):
            c+=1
print(c)
```

```python
# 189
f = open('24-181.txt').readline()
f = f.replace('A', 'Y')
f = f.replace('E', 'Y')
f = f.replace('I', 'Y')
f = f.replace('O', 'Y')
f = f.replace('U', 'Y')


k=0
v = 0
l = 0
b = 0
for i in range(len(f)):
    if f[i]=='Y':
        v+=1
    if f[i]=='.':
        b+=1
    while v>7 or b>0:
        if f[l]=='Y':
            v-=1
        if f[l]=='.':
            b-=1
        l+=1
    if (i-l)>k:
        k=i-l
print(k+1)
```
