```python
f= open('17_2.txt').readlines()
res=[]
a = list(map(int,f))
for i in range(0,len(a)-2):
    r = sorted([a[i], a[i+1], a[i+2]])
    if max(r)**2<(r[0]**2+r[1]**2) and r[2]<(r[1]+r[0]):
        res.append(sum(r))
print(len(res),max(res))
```
