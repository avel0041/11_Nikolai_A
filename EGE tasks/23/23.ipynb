```python
def f(s,e):
    if s==e:
        return 1
    if s<e:
        return 0
    if int(str(s)[-2])>int(str(s)[-1]):
        z=int(str(s)[:-2]+str(s)[-1]+str(s)[-2])
        return f(s-3,e)+f(z,e)
    else:
        return f(s-3,e)
    
    
print(f(1001,959)*f(959,902))
```

```python
def f(s,e):
    if s>e:
        return 0
    if s==e:
        return 1
    if s==11:
        return 0
    return f(s+1,e) +f(s*2, e)

def g(s,e):
    if s>e:
        return 0
    if s==e:
        return 1
    if s==12:
        return 0
    return g(s+1,e) +g(s*2, e)
print(g(2,11)*g(11,24) + f(2,12)*f(12,24))
```

```python
def f(s, e, v):
    if s>e:
        return 0
    if s==e:
        return 1
    if ('11' in v) or ('22' in v):
        return 0
    return f(s+1, e, v+'1') + f(s+2, e, v+'1') + f(s*2,e,v+'2') + f(s*3, e,v+'2') 

print(f(1, 24, ''))
```
