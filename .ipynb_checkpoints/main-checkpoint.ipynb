```python
def f(x):
    res=[]
    for n in range(2,int(x**0.5)+1):
        if x%n==0:
            res.append(n)
            res.append(x//n)
            break
    if len(res)==0:
        return 0
    return max(res)+min(res)

k=0
for n in range(800_000,100_000_000):
    b=f(n)
    if b%10==4:
        k+=1
        print(n,b)
        if k==5:
            break
```

```python
f=open('24_1064.txt').readlines()
res=[]
for s in f:
    k=0
    k_max=0
    z = []
    for i in range(len(s)-1):
        if s[i]==s[i+1]:
            k+=1
            if k == k_max:
                z.append(s[i])
            if k>k_max:
                z = [s[i]]
                k_max=k
        else:
            k=0
    z = set(z)
    z = list(z)
    res = res + z
for el in 'ABCDEF':
    print(el + ': ' + str(res.count(el)))

```
